#### Ejercicio 1
 1. Separar X, de y
 2. Crear un pipeline: Convertir variables categoricas en numericas con one-hot-encoding, normalizar variables numericas con standardScaler

````python
# Define features and target variable
X = tips.drop('total_bill', axis=1)
y = tips['total_bill']

# Identificar variables numéricas y categóricas
````

````python
# Preprocesamiento

# Identificar variables numéricas y categóricas
numeric_features = X.select_dtypes(include=['number']).columns.tolist()
categorical_features = X.select_dtypes(include=['object', 'category']).columns.tolist()

# Pipeline: one-hot encoding para categóricas y estandarización para numéricas
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(drop='first'), categorical_features)
    ]
)
````

## Actividad: Validación Cruzada con Regresión Lineal

En esta actividad pondrás en práctica la técnica de **validación cruzada (K-Fold Cross-Validation)** para evaluar el desempeño de un modelo de regresión lineal. La validación cruzada es una estrategia que permite dividir los datos en varias particiones (o *folds*), entrenar el modelo en una parte y evaluarlo en la otra, repitiendo el proceso varias veces para obtener una medida más robusta del error del modelo.

### Instrucciones

1. **Configura la validación cruzada** usando `KFold` con 5 particiones (*n\_splits=5*), habilitando `shuffle=True` para mezclar los datos y fijando `random_state=42` para asegurar reproducibilidad.

2. **Inicializa un modelo de regresión lineal** con `LinearRegression()` de *scikit-learn*.

3. **Divide los datos en entrenamiento y validación** para cada fold que genere `KFold`.

   * Usa los índices de `train_index` y `val_index` para separar `X_train, X_val` y `y_train, y_val`.

4. **Entrena el modelo**  en los datos de entrenamiento de cada fold.

5. **Genera predicciones** sobre los datos de validación.

6. **Calcula el error** en cada fold usando la métrica **Error Cuadrático Medio (MSE)** con `mean_squared_error`.

7. **Imprime los resultados** mostrando el número de fold y el MSE correspondiente.

### 📌 Resultado esperado

Al finalizar, deberías ver en consola algo como:

```
Fold 1 - MSE: ...
Fold 2 - MSE: ...
Fold 3 - MSE: ...
Fold 4 - MSE: ...
Fold 5 - MSE: ...
```



````python
# Validación cruzada - Regresión Lineal

# Configuración de la validación cruzada con 5 particiones
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# Modelo de regresión lineal combinado con el preprocesamiento
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

mse_scores = []
````

````python
# Iteración sobre cada partición generada por KFold
for fold, (train_index, val_index) in enumerate(kf.split(X), start=1):
    X_train, X_val = X.iloc[train_index], X.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)

    mse = mean_squared_error(y_val, y_pred)
    mse_scores.append(mse)
    print(f"Fold {fold} - MSE: {mse:.4f}")
````

````python
# Resumen del error de la validación cruzada
print(f"\nMSE promedio: {np.mean(mse_scores):.4f}")
print(f"Desviación estándar del MSE: {np.std(mse_scores):.4f}")
````

#### Regularización
Aplicar modelos de regresión lineal con regularización (Ridge y Lasso) al dataset Tips para predecir la variable **total_bill**, identificando qué variables explicativas son más relevantes y comparando los efectos de la regularización sobre los coeficientes.

1. Seleccionar el mejor alpha usando CV
2. Validar de nuevo el rendimiento del regresor usando el modelo ajustado
3. Hacer el proceso para Lasso y Ridge

````python
# Lasso Regression (selección de variables)

# Transformar las variables con el preprocesador
X_trans = preprocessor.fit_transform(X)
feature_names = preprocessor.get_feature_names_out()

# Seleccionar el mejor alpha mediante validación cruzada
lasso_cv = LassoCV(alphas=np.logspace(-3, 2, 100), cv=5, random_state=42)
lasso_cv.fit(X_trans, y)

# Evaluar rendimiento con el alpha óptimo
lasso_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', Lasso(alpha=lasso_cv.alpha_, max_iter=5000))
])
lasso_scores = cross_val_score(lasso_model, X, y, cv=kf, scoring='neg_mean_squared_error')
lasso_rmse = np.sqrt(-lasso_scores)

print(f"Mejor alpha (Lasso): {lasso_cv.alpha_:.4f}")
print(f"RMSE por fold (Lasso): {np.round(lasso_rmse, 4)}")
print(f"RMSE promedio (Lasso): {lasso_rmse.mean():.4f}")
````

````python
# Ridge Regression (comparación)

# Seleccionar el mejor alpha mediante validación cruzada
ridge_cv = RidgeCV(alphas=np.logspace(-3, 2, 100), cv=5)
ridge_cv.fit(X_trans, y)

# Evaluar rendimiento con el alpha óptimo
ridge_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', Ridge(alpha=ridge_cv.alpha_))
])
ridge_scores = cross_val_score(ridge_model, X, y, cv=kf, scoring='neg_mean_squared_error')
ridge_rmse = np.sqrt(-ridge_scores)

print(f"Mejor alpha (Ridge): {ridge_cv.alpha_:.4f}")
print(f"RMSE por fold (Ridge): {np.round(ridge_rmse, 4)}")
print(f"RMSE promedio (Ridge): {ridge_rmse.mean():.4f}")
````

````python
# Comparación de coeficientes: relevancia de las variables

coef_df = pd.DataFrame({
    'variable': feature_names,
    'Lasso': lasso_cv.coef_,
    'Ridge': ridge_cv.coef_
})
coef_df['|Lasso|'] = coef_df['Lasso'].abs()
coef_df = coef_df.sort_values('|Lasso|', ascending=False)

coef_df[['variable', 'Lasso', 'Ridge']].head(10)
````

**Interpretación del Caso 1.** El modelo lineal sin regularizar obtuvo un MSE promedio de 35.74 en validación cruzada, equivalente a un RMSE cercano a 5.98. La regularización Lasso (alpha óptimo 0.1322) alcanzó un RMSE promedio de 5.97, mientras que Ridge (alpha óptimo 13.85) obtuvo 5.96. Los tres desempeños resultan prácticamente equivalentes, lo cual es esperable en un conjunto de datos pequeño y con baja colinealidad; sin embargo, Lasso aporta valor interpretativo al anular los coeficientes de las variables menos relevantes y señalar cuáles explican realmente el total de la cuenta.

* Clasificación binaria (Maligno vs. Benigno).
  

````python
np.random.seed(42)
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df.head(3)
````

## Ejercicio 1

Hacer un experimento inicial para establecer el baseline:
1. Separar X, de y
2. Crear cojuntos de entrnamiento y prueba : 70%-30%. Usar la siguiente configuración: train_test_split(x, y, test_size=0.30, stratify=y, random_state=42)
3. Usar normalizador: StandardScaler
4. Entrenar un modelo de regresión logistica sin ninguna penalidad, evaluar rendimiento


````python
# 1. Separar variables predictoras y variable objetivo
X = df.drop('target', axis=1)
y = df['target']

# 2. División en entrenamiento y prueba (70% - 30%) con estratificación
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=42
)
````

````python
# 3. Normalización de las variables
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Modelo de regresión logística sin penalización
log_model = LogisticRegression(penalty=None)
#  Ajustar el modelo con el conjunto de entrenamiento normalizado
log_model.fit(X_train_scaled, y_train)

# Predicciones sobre el conjunto de prueba
y_pred = log_model.predict(X_test_scaled)
````

````python
# Exactitud del modelo base (baseline)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy del baseline (sin regularización): {accuracy:.4f}")
````

````python
# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=data.target_names, yticklabels=data.target_names)
plt.xlabel('Etiqueta predicha')
plt.ylabel('Etiqueta real')
plt.title('Matriz de confusión - Regresión Logística (baseline)')
plt.show()
````

````python
# Reporte de clasificación
print(classification_report(y_test, y_pred, target_names=data.target_names))
````

### Ejercicio 2

Usar Validación cruzada estratificada para corroborar los resultados anteriores. Imprimir el rendimiento en cada fold. Calcular el promedio

````python
# Validación cruzada estratificada para corroborar el baseline
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

log_reg = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=5000, penalty=None))
])

accuracies = []
for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]
    log_reg.fit(X_tr, y_tr)
    acc = log_reg.score(X_val, y_val)
    accuracies.append(acc)
    print(f"Fold {fold} - Accuracy: {acc:.4f}")

print(f"\nAccuracy promedio: {np.mean(accuracies):.4f}")
````

````python
# 1. Cargar dataset
X = df.drop('target', axis=1)
y = df['target']

# 2. Validación cruzada estratificada
# Definimos pipeline: escalado + regresión logística
log_reg = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=5000, penalty=None))  # sin regularización
])

# 3. Definir validación cruzada estratificada
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 4. Evaluar con accuracy
scores = cross_val_score(log_reg, X, y, cv=cv, scoring="accuracy")

print("Validación cruzada - Regresión Logística (sin regularización)")
print("Accuracy por fold:", np.round(scores, 4))
print("Accuracy promedio:", np.mean(scores))
````

### Ejercicio 3

2. Reentrenar y validar de nuevo usando solamente las variables seleccionadas

````python
#  Aplicar L1 (Lasso) para selección de variables

# Modelo con regularización L1 (Lasso)
lasso_model = LogisticRegression(
    penalty="l1", solver="saga", max_iter=5000, C=1.0
)

# Escalamos antes de entrenar
scaler_lasso = StandardScaler()
X_scaled = scaler_lasso.fit_transform(X)

# Entrenamos con todas las variables
lasso_model.fit(X_scaled, y)

# Identificamos las variables cuyo coeficiente es distinto de cero
coef = pd.Series(lasso_model.coef_[0], index=X.columns)
selected_features = coef[coef != 0].index.tolist()

print("\nVariables seleccionadas por Lasso:")
print(coef[coef != 0])
````

````python
#  Reentrenar usando solo variables seleccionadas

X_selected = X[selected_features]

log_reg_selected = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=5000, penalty=None))
])

scores_selected = cross_val_score(log_reg_selected, X_selected, y, cv=cv, scoring="accuracy")

print("\nValidación cruzada - Modelo con variables seleccionadas por Lasso")
print("Accuracy por fold:", scores_selected)
print("Accuracy promedio:", np.mean(scores_selected))
````

````python
# Cargar datos
titanic_df = sns.load_dataset('titanic')
titanic_df.head(3)
````

````python
# Selección de variables útiles

# Se descartan las variables que generan fuga de información o son redundantes:
#   - 'alive' es equivalente al target 'survived'
#   - 'class' y 'embark_town' son versiones categóricas de 'pclass' y 'embarked'
#   - 'deck', 'who', 'adult_male' y 'alone' aportan información redundante o presentan muchos faltantes
features = ['pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked']
titanic = titanic_df[features + ['survived']].copy()

titanic.head()
````

````python
# Revisión de valores faltantes en las variables seleccionadas
titanic.isnull().sum()
````

````python
# Identificar variables numéricas y categóricas
numeric_features = ['age', 'sibsp', 'parch', 'fare']
categorical_features = ['pclass', 'sex', 'embarked']

# Preprocesamiento
#   - Numéricas: imputación con la mediana y estandarización
#   - Categóricas: imputación con la moda y one-hot encoding
preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), numeric_features),
        ('cat', Pipeline([
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(drop='first'))
        ]), categorical_features)
    ]
)

X = titanic.drop('survived', axis=1)
y = titanic['survived']
````

````python
# División en entrenamiento y prueba (80% - 20%) con estratificación
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print(f"Entrenamiento: {X_train.shape[0]} muestras")
print(f"Prueba: {X_test.shape[0]} muestras")
````

````python
# ===============================
# Validación cruzada con Regresión Logística (sin regularización)
# ===============================

log_reg = Pipeline([
    ("preprocessor", preprocessor),
    ("clf", LogisticRegression(max_iter=5000, penalty=None))
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(log_reg, X, y, cv=cv, scoring="accuracy")

print("Validación cruzada - Regresión Logística (sin regularización)")
print("Accuracy por fold:", np.round(scores, 4))
print("Accuracy promedio:", np.mean(scores))
````

````python
# ===============================
# Aplicar LASSO (L1) para selección de variables
# ===============================

# Transformar las variables (imputación + one-hot + escalado)
X_trans = preprocessor.fit_transform(X)
feature_names = preprocessor.get_feature_names_out()

lasso_model = LogisticRegression(
    penalty="l1", solver="saga", max_iter=5000, C=1.0
)
lasso_model.fit(X_trans, y)

# Extraer coeficientes después de one-hot
coef = pd.Series(lasso_model.coef_[0], index=feature_names)
selected_features = coef[coef != 0].index.tolist()

print("Variables seleccionadas por Lasso:")
print(coef[coef != 0])
````

````python
# Reentrenar con validación cruzada usando las variables seleccionadas por Lasso

mask = np.isin(feature_names, selected_features)
X_selected = X_trans[:, mask]

log_reg_selected = LogisticRegression(max_iter=5000, penalty=None)
scores_selected = cross_val_score(log_reg_selected, X_selected, y, cv=cv, scoring="accuracy")

print("\nValidación cruzada - Modelo con variables seleccionadas por Lasso")
print("Accuracy por fold:", np.round(scores_selected, 4))
print("Accuracy promedio:", np.mean(scores_selected))
````
