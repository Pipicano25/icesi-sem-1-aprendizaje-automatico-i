## **Caso 1: Horas de Estudio y Calificaciones (Regresión Lineal Simple)**

**Dataset:** `score_updated.csv`  
**Relación esperada:** Lineal  

### **Parte 1: Verificación de Supuestos**
1. Cargar los datos y visualizar la relación entre horas de estudio y calificaciones mediante un **gráfico de dispersión**.  
2. Evaluar la distribución de las variables usando histogramas y boxplots.  
---


```python
# Carga de librerías y configuración general del entorno
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy import stats

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

# Directorio donde se ubican los archivos de datos.
# Ajustar la ruta si los CSV se encuentran en otra ubicación.
DATA_DIR = "../../Sesión 1/Data - Sesión 1-20251215/"
```

```python
# Caso 1 - Parte 1: Verificación de supuestos
calificaciones = pd.read_csv(DATA_DIR + "score_updated.csv")
print("Dimensiones:", calificaciones.shape)
print(calificaciones.head())

# 1. Relación entre horas de estudio y calificaciones
plt.figure(figsize=(8, 5))
sns.scatterplot(data=calificaciones, x="Hours", y="Scores")
plt.title("Relación entre horas de estudio y calificaciones")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificación")
plt.show()

# 2. Distribución de las variables mediante histogramas
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(calificaciones["Hours"], kde=True, ax=axes[0])
axes[0].set_title("Distribución de las horas de estudio")
sns.histplot(calificaciones["Scores"], kde=True, ax=axes[1])
axes[1].set_title("Distribución de las calificaciones")
plt.tight_layout()
plt.show()

# 2. Distribución de las variables mediante boxplots
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(y=calificaciones["Hours"], ax=axes[0])
axes[0].set_title("Boxplot de horas de estudio")
sns.boxplot(y=calificaciones["Scores"], ax=axes[1])
axes[1].set_title("Boxplot de calificaciones")
plt.tight_layout()
plt.show()
```

### **Parte 2: Ajuste del Modelo y Evaluación**
1. Ajustar una **regresión lineal** usando `statsmodels`.  
2. Interpretar los coeficientes del modelo y el **summary()**:
   - ¿Cuál es el valor de **R²**? ¿Qué indica sobre el modelo?  
   - ¿Los coeficientes son estadísticamente significativos?  
   - ¿Cómo interpretamos el coeficiente de la variable "horas de estudio"?  

3. Verificar los supuestos de la regresión lineal:
   - **Linealidad:** ¿La relación entre las variables parece una línea recta?  
   - **Normalidad de los residuos:** Graficar un **histograma de los residuos** y realizar un **test de normalidad**.  
   - **Homoscedasticidad:** Generar un **gráfico de residuos vs valores ajustados**.  
4. Responder: ¿Es apropiado un modelo de regresión lineal en este caso? ¿Por qué?  


---


```python
# Caso 1 - Parte 2: Ajuste del modelo de regresión lineal
X = sm.add_constant(calificaciones["Hours"])
y = calificaciones["Scores"]

modelo_caso1 = sm.OLS(y, X).fit()
print(modelo_caso1.summary())

# Verificación de los supuestos de la regresión lineal
residuos = modelo_caso1.resid
valores_ajustados = modelo_caso1.fittedvalues

# Normalidad de los residuos
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(residuos, kde=True, ax=axes[0])
axes[0].set_title("Histograma de los residuos")
sm.qqplot(residuos, line="45", ax=axes[1])
axes[1].set_title("Q-Q plot de los residuos")
plt.tight_layout()
plt.show()

estadistico, p_valor = stats.shapiro(residuos)
print(f"Test de normalidad de Shapiro-Wilk: estadístico={estadistico:.4f}, p-valor={p_valor:.4f}")

# Homoscedasticidad: residuos frente a valores ajustados
plt.figure(figsize=(8, 5))
sns.scatterplot(x=valores_ajustados, y=residuos)
plt.axhline(0, color="red", linestyle="--")
plt.title("Residuos vs valores ajustados")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.show()
```

El modelo ajustado presenta un coeficiente de determinación R² de 0.9814, lo que significa que las horas de estudio explican aproximadamente el 98% de la variabilidad observada en las calificaciones. El coeficiente asociado a las horas de estudio (9.90) es estadísticamente significativo, con un p-valor del orden de 10⁻⁸³, muy por debajo del umbral convencional de 0.05, lo que permite descartar con confianza la hipótesis de que su efecto sea nulo. En términos prácticos, cada hora adicional de estudio se asocia con un incremento medio de cerca de 9.9 puntos en la calificación. La relación entre ambas variables se aproxima bien a una línea recta y los residuos no muestran patrones sistemáticos, por lo que la regresión lineal simple resulta apropiada para este caso.


### **Parte 3: Evaluación del Modelo con RMSE**
1. Calcular el **RMSE**.  
2. Reflexionar:  
   - ¿El RMSE es bajo o alto?  
   - ¿Qué indica esto sobre la precisión del modelo?  
   - ¿Cómo podríamos mejorar la predicción?  

---


```python
# Caso 1 - Parte 3: Evaluación del modelo con RMSE
predicciones = modelo_caso1.predict(X)
rmse_caso1 = np.sqrt(np.mean((y - predicciones) ** 2))
print(f"RMSE del modelo: {rmse_caso1:.4f}")
```

El RMSE del modelo es de 3.40 puntos sobre una escala de calificaciones que va de 12 a 99, lo que representa un error relativo pequeño y confirma la buena capacidad predictiva del modelo. Un RMSE bajo indica que las predicciones se desvían poco de los valores observados. Para mejorar aún más la predicción se podría incorporar información adicional del estudiante, como la asistencia o el desempeño previo, o recurrir a una transformación de las variables si la relación dejara de ser lineal. No obstante, dado el ajuste alcanzado, las ganancias potenciales de añadir complejidad serían marginales.

### **Parte 5: Reevaluación del Modelo**  

1. **Volver a ajustar la regresión lineal** con los datos modificados.

```python
# Caso 1 - Parte 5: Reevaluación del modelo con los datos modificados
X_mod = sm.add_constant(calificaciones_mod["Hours"])
y_mod = calificaciones_mod["Scores"]

modelo_mod = sm.OLS(y_mod, X_mod).fit()
print(modelo_mod.summary())

predicciones_mod = modelo_mod.predict(X_mod)
rmse_mod = np.sqrt(np.mean((y_mod - predicciones_mod) ** 2))
print(f"RMSE tras añadir los valores atípicos: {rmse_mod:.4f}")
```

Este comportamiento evidencia la gran sensibilidad de la regresión lineal a los valores atípicos: al minimizar la suma de los errores al cuadrado, unos pocos puntos extremos arrastran la recta de ajuste y destruyen una relación que antes era clara. Para mitigar su impacto se pueden emplear métodos robustos, como la regresión robusta o la minimización del error absoluto. La decisión de eliminar o conservar un valor atípico debe basarse en su naturaleza: si corresponde a un error de captura conviene corregirlo o retirarlo, mientras que si representa una observación real y relevante debe mantenerse y, en su caso, modelarse de forma específica.

## **Caso 2: Temperatura y Ventas de Helado (Relación No Lineal)**

**Dataset:** `Ice_cream_selling_data.csv`  
**Relación esperada:** No lineal  

### **Parte 1: Exploración de Datos y Supuestos**
1. Cargar los datos y visualizar la relación mediante un **gráfico de dispersión**.  
2. Evaluar la distribución de las variables usando histogramas y boxplots.

```python
# Caso 2 - Parte 1: Exploración de datos y supuestos
helado = pd.read_csv(DATA_DIR + "Ice_cream_selling_data.csv")
helado.columns = ["Temperatura", "Ventas"]
print("Dimensiones:", helado.shape)
print(helado.head())

# Relación entre temperatura y ventas
plt.figure(figsize=(8, 5))
sns.scatterplot(data=helado, x="Temperatura", y="Ventas")
plt.title("Relación entre temperatura y ventas de helado")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Ventas de helado (unidades)")
plt.show()

# Distribución de las variables
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(helado["Temperatura"], kde=True, ax=axes[0])
axes[0].set_title("Distribución de la temperatura")
sns.histplot(helado["Ventas"], kde=True, ax=axes[1])
axes[1].set_title("Distribución de las ventas")
plt.tight_layout()
plt.show()
```

El gráfico de dispersión entre la temperatura y las ventas de helado no muestra una tendencia lineal clara: la nube de puntos presenta una forma dispersa sin una dirección definida y la correlación entre ambas variables es débil (alrededor de -0.18). Este comportamiento sugiere que la relación entre la temperatura y las ventas no es lineal y que un modelo de regresión lineal simple probablemente no capturará de forma adecuada la estructura subyacente de los datos.

### **Parte 2: Ajuste del Modelo y Evaluación**
1. Ajustar una **regresión lineal** usando `statsmodels`.  
2. Interpretar los coeficientes del modelo y el **summary()**:
   - ¿Cuál es el valor de **R²**? ¿Qué indica sobre el modelo?  
   - ¿Los coeficientes son estadísticamente significativos?  
   - ¿Cómo interpretamos el coeficiente de la variable "temperatura"?

3. Verificar los supuestos de la regresión lineal:
   - **Linealidad:** ¿La relación entre las variables parece una línea recta?  
   - **Normalidad de los residuos:** Graficar un **histograma de los residuos** y realizar un **test de normalidad**.  
   - **Homoscedasticidad:** Generar un **gráfico de residuos vs valores ajustados**.  


---

```python
# Caso 2 - Parte 2: Ajuste del modelo de regresión lineal
X2 = sm.add_constant(helado["Temperatura"])
y2 = helado["Ventas"]

modelo_caso2 = sm.OLS(y2, X2).fit()
print(modelo_caso2.summary())

# Verificación de supuestos: normalidad de los residuos
residuos2 = modelo_caso2.resid
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(residuos2, kde=True, ax=axes[0])
axes[0].set_title("Histograma de los residuos")
sm.qqplot(residuos2, line="45", ax=axes[1])
axes[1].set_title("Q-Q plot de los residuos")
plt.tight_layout()
plt.show()

# Homoscedasticidad: residuos frente a valores ajustados
plt.figure(figsize=(8, 5))
sns.scatterplot(x=modelo_caso2.fittedvalues, y=residuos2)
plt.axhline(0, color="red", linestyle="--")
plt.title("Residuos vs valores ajustados")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.show()
```

El ajuste lineal confirma la sospecha planteada durante la exploración: el R² del modelo es de apenas 0.03, de modo que la temperatura explica menos del 4% de la variabilidad de las ventas, y el coeficiente asociado no resulta estadísticamente significativo. La relación entre la temperatura y las ventas de helado no puede describirse mediante una recta, lo que justifica el uso de un modelo no lineal, como la regresión polinomial, capaz de capturar la curvatura presente en los datos.

### **Parte 3: Evaluación del Modelo con RMSE**
1. Calcular el **RMSE**.  
2. Reflexionar:  
   - ¿Que podemos concluir al usar solamente el RMSE como medida de desempeño?  
   - ¿Es posible hacer ajustes para mejorar el modelo?  

---


```python
# Caso 2 - Parte 3: Evaluación del modelo con RMSE
predicciones2 = modelo_caso2.predict(X2)
rmse_caso2 = np.sqrt(np.mean((y2 - predicciones2) ** 2))
print(f"RMSE del modelo lineal: {rmse_caso2:.4f}")
print(f"R² del modelo lineal: {modelo_caso2.rsquared:.4f}")
```

Utilizar únicamente el RMSE como medida de desempeño ofrece una visión incompleta del modelo: un RMSE de aproximadamente 11.95 unidades puede parecer razonable en términos absolutos, pero ocultaría el hecho de que el modelo lineal explica una fracción mínima de la variabilidad, con un R² de 0.03. Por ello, el RMSE debe acompañarse de métricas como el R² y de un análisis gráfico de los residuos para juzgar la calidad real del ajuste. El desempeño puede mejorarse sustancialmente recurriendo a un modelo no lineal, como la regresión polinomial, que se ajuste a la curvatura de los datos.


## **Caso 3: Análisis de Enfermedades Cardíacas (Selección de Variables y Comparación de Modelos)**

**Dataset:** `dataset_2190_cholesterol.csv`  
**Objetivo:** Identificar variables clave y construir modelos predictivos  

### **Parte 1: Identificación de Variables**
1. Cargar el dataset y explorar sus columnas.  
2. Identificar la **variable objetivo** (num: presencia de enfermedad cardíaca).  
3. Calcular la **correlación** de todas las variables con "num" y ordenarlas de mayor a menor.  

---


```python
# Caso 3 - Parte 1: Identificación de variables
cardiaco = pd.read_csv(DATA_DIR + "dataset_2190_cholesterol.csv")
print("Dimensiones:", cardiaco.shape)
print(cardiaco.head())
print("\nTipos de datos por columna:")
print(cardiaco.dtypes)
print("\nValores nulos por columna:")
print(cardiaco.isnull().sum())
```

```python
# Correlación de todas las variables numéricas con la variable objetivo "num"
columnas_numericas = cardiaco.select_dtypes(include=np.number).columns.tolist()
correlacion_num = cardiaco[columnas_numericas].corr()["num"].drop("num")

correlacion_ordenada = correlacion_num.sort_values(key=abs, ascending=False)
print("Correlación de cada variable con 'num' (ordenadas por magnitud):")
print(correlacion_ordenada)
```

### **Parte 2: Selección de Variables y Modelado**
1. Seleccionar las **tres variables con mayor correlación** con "num".  
2. Ajustar **tres modelos de regresión lineal** (uno para cada variable independiente).  
3. Evaluar los coeficientes y métricas del summary():
   - **R² ajustado**  
   - **Significancia de los coeficientes**  
     


```python
# Caso 3 - Parte 2: Selección de variables y modelado
variables_top = correlacion_ordenada.index[:3].tolist()
print("Tres variables con mayor correlación con 'num':", variables_top)

modelos_caso3 = {}
for variable in variables_top:
    X3 = sm.add_constant(cardiaco[variable])
    y3 = cardiaco["num"]
    modelos_caso3[variable] = sm.OLS(y3, X3).fit()
    print("\n--- Modelo:", variable, "---")
    print(modelos_caso3[variable].summary())
```

```python
# Resumen comparativo de los tres modelos: R² ajustado y significancia
comparacion = pd.DataFrame({
    "Variable": variables_top,
    "R2": [modelos_caso3[v].rsquared for v in variables_top],
    "R2_ajustado": [modelos_caso3[v].rsquared_adj for v in variables_top],
    "Coeficiente": [modelos_caso3[v].params[v] for v in variables_top],
    "p_valor": [modelos_caso3[v].pvalues[v] for v in variables_top]
})
print(comparacion.to_string(index=False))
```

De las variables analizadas, oldpeak (la depresión del segmento ST en el electrocardiograma) es la que presenta el mayor poder predictivo individual, con un R² de 0.25, seguida de thalach (frecuencia cardíaca máxima, con correlación negativa) y cp (tipo de dolor torácico). Aunque cada variable aporta información de forma individual, los R² obtenidos son moderados, lo que indica que la combinación de varias variables en un modelo múltiple podría mejorar el desempeño. En conjunto, estos resultados refuerzan la idea de que la enfermedad cardíaca es un fenómeno multifactorial, en el que el estado funcional del corazón y el tipo de dolor torácico resultan más informativos que el nivel de colesterol considerado de forma aislada.
