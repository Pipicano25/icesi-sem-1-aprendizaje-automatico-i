### 📌 **Diferencia entre Correlación y Causalidad**  
📖 **Definición:**  
La **correlación** mide el grado de relación entre dos variables, mientras que la **causalidad** implica que una variable es responsable de los cambios en otra.  

🔹 **Ejemplo ilustrativo:**  
- Se observa una alta correlación entre el consumo de helado y el número de ataques de tiburón.  
- ¿Significa esto que comer helado provoca ataques de tiburón? ❌ **No**.  
- La verdadera causa es el verano: cuando hace calor, más personas compran helado y más personas nadan en el mar, aumentando la probabilidad de ataques.

### 📌 **Tipos de Correlación**  
🔹 **Correlación positiva (+)** → Aumenta una variable, aumenta la otra.  
🔹 **Correlación negativa (-)** → Aumenta una variable, disminuye la otra.  
🔹 **Correlación nula (≈0)** → No hay relación clara entre las variables.  

📊 **Ejemplos gráficos:**  
1. **Positiva**: Ingreso mensual vs. gasto en entretenimiento.  
2. **Negativa**: Número de horas de ejercicio vs. nivel de colesterol.  
3. **Nula**: Tamaño del zapato vs. calificaciones académicas.


### 📌 **Coeficiente de Correlación de Pearson**  
📖 **Definición matemática:**  
$
r = \frac{\sum (X_i - \bar{X}) (Y_i - \bar{Y})}{\sqrt{\sum (X_i - \bar{X})^2} \sqrt{\sum (Y_i - \bar{Y})^2}}
$

📊 **Interpretación de valores:**  
- **r = 1** → Correlación perfectamente positiva.  
- **r = -1** → Correlación perfectamente negativa.  
- **r = 0** → No hay relación lineal.  

📌 **Limitaciones:**  
- No detecta relaciones no lineales.  
- Es sensible a valores atípicos.  

📊 **Ejemplo gráfico:**  

-----

```python
import pandas as pd
import seaborn as sns
df = pd.DataFrame({'A': [1, 2, 3, 4, 5], 
                   'B': [2, 4, 6, 8, 10], 
                   'C': [5, 3, 8, 2, 7]})

correlation_matrix = df.corr()
print(correlation_matrix)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generar datos
np.random.seed(0)
x = np.linspace(0, 10, 100)

# Correlación positiva
y_positive = 2 * x + np.random.normal(0, 2, len(x))

# Correlación negativa
y_negative = -2 * x + np.random.normal(0, 2, len(x))

# Correlación cercana a cero (distribución aleatoria)
y_zero = np.random.normal(5, 2, len(x))

# Crear subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Gráfico de correlación positiva
axes[0].scatter(x, y_positive, color='green')
axes[0].set_title("Correlación Positiva")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

# Gráfico de correlación negativa
axes[1].scatter(x, y_negative, color='red')
axes[1].set_title("Correlación Negativa")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")

# Gráfico de correlación cero
axes[2].scatter(x, y_zero, color='blue')
axes[2].set_title("Correlación Cero")
axes[2].set_xlabel("X")
axes[2].set_ylabel("Y")

plt.tight_layout()
plt.show()
```

```python
# 3. Interpretación básica
if correlacion > 0.7:
    interpretacion = "Fuerte correlación positiva"
elif correlacion > 0.3:
    interpretacion = "Correlación positiva moderada"
elif correlacion > 0:
    interpretacion = "Correlación positiva débil"
elif correlacion < -0.7:
    interpretacion = "Fuerte correlación negativa"
elif correlacion < -0.3:
    interpretacion = "Correlación negativa moderada"
elif correlacion < 0:
    interpretacion = "Correlación negativa débil"
else:
    interpretacion = "Sin correlación"

print("Interpretación:", interpretacion)
```

### 📖 **Caso: Temperatura y Ventas de Helado**  
📂 **Datos:** Temperatura diaria y ventas de helado en una heladería: `Ice_cream_selling_data.csv`

**Ice Cream Selling dataset**: El conjunto de datos muestra la relación entre la temperatura y las ventas de helados. Sirve como un ejemplo práctico para comprender e implementar la regresión polinómica, una técnica poderosa para modelar relaciones no lineales en los datos.

📌 **Preguntas:**  
- ¿Cómo cambia la correlación en distintas estaciones?

```python
import pandas as pd
import seaborn as sns
import os
# 
ruta_archivo = os.getcwd()

df = pd.read_csv(os.path.join(ruta_archivo, '..', 'Data - Sesión 1-20250729', 'Ice_cream_selling_data.csv'))
df.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Gráfico de dispersión
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Temperature (°C)", y="Ice Cream Sales (units)")
plt.title("Temperatura vs. Ventas de Helado")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Ventas")
plt.grid(True)
plt.show()

# 2. Calcular correlación lineal
r = df['Temperature (°C)'].corr(df['Ice Cream Sales (units)'])
print(f"Coeficiente de correlación lineal (r): {r:.2f}")
```

¿Por qué la correlación $r$ puede ser baja aunque haya relación?
La correlación de Pearson solo detecta relaciones lineales.

Si la relación es cuadrática (por ejemplo, las ventas suben con la temperatura hasta cierto punto y luego bajan), el valor de $r$ puede estar cerca de 0.

Esto no significa que no haya relación, sino que no es lineal.

```python
# Detección visual de outliers para valores atipicos
sns.boxplot(x=df['Ice Cream Sales (units)'])
plt.title("Outliers en Ventas")
plt.show()
```

Comparativa: Pearson vs. Kendall vs. Spearman

| Característica                 | **Pearson**                    | **Spearman (ρ)**                             | **Kendall (τ)**                               |
| ------------------------------ | ------------------------------ | -------------------------------------------- | --------------------------------------------- |
| **Tipo de relación detectada** | **Lineal**                     | **Monótona (no necesariamente lineal)**      | **Monótona (por orden)**                      |
| **Tipo de datos**              | Continuos, numéricos           | Ordinales o continuos (convertidos a rangos) | Ordinales, categóricos ordenados, rangos      |
| **Sensibilidad a outliers**    | Alta                           | Baja                                         | Muy baja                                      |
| **Interpretación**             | Cambios proporcionales         | Cambios en el orden                          | Proporción de concordancias vs. discordancias |
| **Escala de resultado**        | de -1 a 1                      | de -1 a 1                                    | de -1 a 1                                     |
| **Uso común**                  | Modelos lineales, regresiones  | Cuando la relación no es lineal              | Comparar ordenamientos                        |
| **Cálculo**                    | Fórmula con medias y varianzas | Basado en rangos                             | Basado en pares ordenados                     |


## Usa Pearson si...
Estás trabajando con variables numéricas continuas.

Esperas una relación lineal.

No hay muchos outliers.

## Usa Spearman si...
Hay una relación creciente o decreciente, pero no necesariamente lineal.

Tienes datos ordinales, o transformados en rangos.

Hay outliers que podrían distorsionar Pearson.

## Usa Kendall si...
Quieres una medida más robusta para muestras pequeñas o con muchos empates.

Te interesa cómo cambian los órdenes relativos.

Buscas mayor estabilidad estadística en el orden.




```python
x="Temperature (°C)"
y="Ice Cream Sales (units)"
```

```python
# Pearson usando pandas
r_pearson = df[x].corr(df[y], method='pearson')
print(f"Pearson: {r_pearson:.3f}")


```

```python
# Spearman usando pandas
r_spearman = df[x].corr(df[y], method='spearman')
print(f"Spearman: {r_spearman:.3f}")
```

```python
# Kendall usando pandas
r_kendall = df[x].corr(df[y], method='kendall')
print(f"Kendall: {r_kendall:.3f}")
```

¿Qué significa eso?
A medida que X sube desde -4 hasta 0, Y baja → relación negativa.

Desde 0 hasta 4, Y sube → relación positiva.

Pero Pearson, Spearman y Kendall están diseñadas para relaciones lineales o monótonas (una dirección).

Por eso, los 3 tipos de correlación te darán un valor bajo o cercano a 0, aunque claramente haya una relación fuerte pero no lineal.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(data=df, x=x, y=y)
plt.title("Relación en forma de U")
plt.xlabel(x)
plt.ylabel(y)
plt.grid(True)
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X = df[[x]]
Y = df[y]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

modelo = LinearRegression()
modelo.fit(X_poly, Y)

# Predicción para graficar
x_range = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)
x_range_poly = poly.transform(x_range)
y_pred = modelo.predict(x_range_poly)

# Gráfico
plt.figure(figsize=(8, 5))
sns.scatterplot(x=X[x], y=Y, label='Datos')
plt.plot(x_range, y_pred, color='red', label='Curva ajustada')
plt.legend()
plt.title("Regresión Cuadrática: Relación en forma de U")
plt.xlabel(x)
plt.ylabel(y)
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
import numpy as np

x="Temperature (°C)"
y="Ice Cream Sales (units)"

X = df[[x]]          # variable independiente (ej. temperatura)
y = df[y]            # variable dependiente (ej. ventas)

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

modelo = LinearRegression()
modelo.fit(X_poly, y)

r2 = modelo.score(X_poly, y)
print(f"Coeficiente de determinación R²: {r2:.3f}")
```
